With Integers (Bitwise OR): Compares numbers at the binary level

```python
# 5 is 0101 in binary, 3 is 0011
print(5 | 3)   # Outputs: 7 (0111 in binary)
print(5 or 3)  # Outputs: 5 (returns the first truthy value)
```

With Sets (Union): Combines unique items from both sets.

```python
print({1, 2} | {2, 3})  # Outputs: {1, 2, 3}
```

With Dictionaries (Merge): Merges two dictionaries together

```python
print({'a': 1} | {'b': 2})  # Outputs: {'a': 1, 'b': 2}
```
